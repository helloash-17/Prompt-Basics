# Talking to AI: Prompt Basics
### A very simple starting guide

## What are we doing here?
You've probably used ChatGPT before — you type a message, it types back an answer. That's what we're going to do here, but using **code** instead of a website, so we can experiment and automate it.

The message you type to the AI is called a **prompt**. Getting good answers from AI is mostly about learning how to write good prompts. That's all "prompt engineering" means — nothing fancy, just: *ask better, get better answers.*

```python
# Step 2: Connect to the AI using your key
#
# We read the key from Colab's Secrets manager (the 🔑 icon on the left)
# instead of typing it in — this is instant and avoids the paste-in box
# sometimes freezing.

from groq import Groq

try:
    from google.colab import userdata
    api_key = userdata.get("GROQ_API_KEY")
except Exception:
    # Not running in Colab (e.g. running locally) — fall back to typing it in.
    import getpass
    api_key = getpass.getpass("Paste your Groq API key here: ")

client = Groq(api_key=api_key)

def ask(prompt):
    """Send a message (prompt) to the AI and return its reply."""
    response = client.chat.completions.create(
        # Groq's fastest free-tier model right now.
        # If this model name ever stops working, check the current list at
        # console.groq.com/docs/models and swap it in here.
        model="openai/gpt-oss-20b",
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content

print("All set! Now we can use ask(\"...\") to talk to the AI.")
```

## 2. Being Vague vs. Being Clear

This is the single most important lesson in prompt engineering: **the AI can only work with what you actually tell it.** A vague question gets a vague, generic answer. A clear, detailed question gets a useful, specific answer.

Below we ask about the *same topic* two different ways. Run both and compare the answers.

```python
# A vague prompt — not much detail about what we actually want
vague_prompt = "Tell me about arrays."

print(ask(vague_prompt))
```

```python
# A clear prompt — says exactly what we want, how long, and in what style
clear_prompt = """Explain what an array is to a first-year engineering student.
Use exactly 3 bullet points, and give one small code example in Python."""

print(ask(clear_prompt))
```

Notice how the second answer is more useful because we told the AI: **who** it's for, **how long** the answer should be, and **what format** to use (bullet points + an example). That's the whole trick — be specific about what you want.

## 3. Show It an Example First

Sometimes it's easier to **show** the AI what you want instead of describing it in words. If you give it one or two examples of the pattern, it will copy that same pattern for your new input.

In the prompt below, look at the pattern: `word -> short meaning`. We give 2 examples, then leave the 3rd one for the AI to fill in the same way.

```python
prompt = """Give a very short, one-line meaning for each word, in the same style as the examples.

Word: Algorithm
Meaning: A step-by-step set of instructions to solve a problem.

Word: Variable
Meaning: A named box in memory that stores a value.

Word: Compiler
Meaning:"""

print(ask(prompt))
```

## 4. Ask It to Explain Step by Step

For anything involving maths or logic, add the words **"think step by step"** to your prompt. This makes the AI work through the problem slowly, which usually gives a more correct final answer — just like showing your working in an exam, instead of jumping straight to the answer.

```python
prompt = """A classroom has 5 rows. Each row has 8 seats. 6 seats are empty.
How many students are sitting in the classroom?

Think step by step, then give the final answer on the last line."""

print(ask(prompt))
```
